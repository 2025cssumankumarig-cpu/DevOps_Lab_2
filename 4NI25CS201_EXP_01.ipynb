```python
import pandas as pd
import numpy as np
```

```python
data={
    "name":['amit','bebo','chetan','amit','divya','ravi'],
    "age":[25,np.nan,30,25,28,22],
    "city":['mysuru','mysuru','bengaluru','mysuru',np.nan,'delhi']
}
df=pd.DataFrame(data)
print("dataset:")
print(df)
```

```python
print("\nshape of dataset(rows,columns):",df.shape)
print("\nColumns names:",df.columns.tolist())
print("\nData types of columns:")
print(df.dtypes)
print("\nDataset info:")
print(df.info())
```

```python
print("\nDescriptive statistics(numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nMode:\n",df.mode(numeric_only=True))
print("\nStandard deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("total missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:",df.duplicated().sum())
print("\nEDA process completed successfully!")
```
